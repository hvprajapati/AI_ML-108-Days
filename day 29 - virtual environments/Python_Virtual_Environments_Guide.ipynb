# Python Virtual Environments Guide
## Stop Breaking Your Python Installation

A **virtual environment** is an isolated Python setup for each project. Without it, different projects fight over package versions. With it, each project has its own clean space.

**Real problem this solves:** Project A needs pandas v1.0, Project B needs pandas v2.0. Without venv -> conflict. With venv -> each project has its own pandas.

## 1. Creating a Virtual Environment

Open your terminal (PowerShell / Command Prompt) and run:

```bash
# Windows
python -m venv myenv

# Mac / Linux
python3 -m venv myenv
```

This creates a folder called `myenv` with an isolated Python inside.

## 2. Activating the Environment

**Windows (PowerShell):**
```bash
myenv\Scripts\Activate.ps1
```

**Windows (Command Prompt):**
```bash
myenv\Scripts\activate.bat
```

**Mac / Linux:**
```bash
source myenv/bin/activate
```

Once activated, you will see `(myenv)` in your terminal prompt. All pip installs now go into this environment.

## 3. Using the Environment

After activation:

```bash
pip install pandas
pip install numpy matplotlib
pip list              # See what's installed

# Save your environment:
pip freeze > requirements.txt

# Later, on another computer:
pip install -r requirements.txt   # Installs everything from the file

# Deactivate when done:
deactivate
```

## 4. What Goes in requirements.txt?

```
numpy==1.24.0
pandas==2.1.0
matplotlib==3.7.0
requests==2.31.0
```

This file is like a receipt -- anyone can recreate your exact environment from it. **Always include this in your projects.**

## 5. The .gitignore File (If Using Git)

Never upload your `myenv/` folder to GitHub. It is huge and platform-specific.

Create a `.gitignore` file with:
```
myenv/
venv/
env/
__pycache__/
*.pyc
```

Your `requirements.txt` is enough -- others can recreate the environment.

## 6. Best Practices

1. **One venv per project** -- don't reuse environments
2. **Always activate before installing** -- check for the `(env_name)` prefix
3. **requirements.txt is mandatory** -- for any project you share
4. **Don't put venv in GitHub** -- add it to .gitignore
5. **Name it something obvious** -- `venv`, `.venv`, or `env`

## 7. Troubleshooting

**"pip is not recognized":** Python is not in your PATH. Reinstall Python and check "Add Python to PATH."

**"Running scripts is disabled" (Windows):** Open PowerShell as Administrator and run:
```bash
Set-ExecutionPolicy -ExecutionPolicy RemoteSigned -Scope CurrentUser
```

**"No module named pip":** Run `python -m ensurepip --upgrade`

# Summary

Virtual environments are not optional -- they are how real Python developers work.

| Command | What It Does |
|---------|-------------|
| `python -m venv myenv` | Create new environment |
| `myenv\Scripts\activate` | Activate (Windows) |
| `source myenv/bin/activate` | Activate (Mac/Linux) |
| `pip install package` | Install in active environment |
| `pip freeze > requirements.txt` | Save package list |
| `pip install -r requirements.txt` | Install from saved list |
| `deactivate` | Exit the environment |

**From now on, start every new project with a virtual environment!**
